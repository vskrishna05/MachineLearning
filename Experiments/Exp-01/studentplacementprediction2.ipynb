```python
df_clean = df.copy()

print(" DATA CLEANING PIPELINE")

print(f"\nStarting shape: {df_clean.shape}")
print(f"Starting missing: {df_clean.isnull().sum().sum()}")
```

```python
print("\nHandling Missing Salary Values")
print("-" * 50)

not_placed = df_clean[df_clean['placement_status'] == 0]
placed = df_clean[df_clean['placement_status'] == 1]

print(f"   Students not placed: {len(not_placed):,}")
print(f"   Students placed: {len(placed):,}")
print(f"   Missing salary values: {df_clean['salary_package_lpa'].isnull().sum():,}")
print(f"   ✓ Missing salaries match non-placement count: {df_clean['salary_package_lpa'].isnull().sum() == len(not_placed)}")
```

```python
print("\n Removing Duplicates")
print("-" * 50)
if df_clean.duplicated().sum() > 0:
    dup_count = df_clean.duplicated().sum()
    df_clean = df_clean.drop_duplicates()
    print(f"Removed {dup_count} duplicate rows")
else:
    print("    No duplicates found")
```

```python
print("\nVerifying Data Types")
print("-" * 50)
categorical_cols = ['branch', 'college_tier']
for col in categorical_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].astype('category')
        print(f"    Set {col} as categorical")
df_clean['placement_status'] = df_clean['placement_status'].astype(int)
print(f"   Set placement_status as integer")

print(f"\nFinal shape: {df_clean.shape}")
print(f"Data retained: {100*len(df_clean)/len(df):.1f}%")

print(" CLEANING COMPLETE....")
```
